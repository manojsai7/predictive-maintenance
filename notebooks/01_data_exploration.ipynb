# Predictive Maintenance - Data Exploration

This notebook demonstrates how to generate synthetic data and explore it for predictive maintenance tasks.

```python
import sys
sys.path.append('../src')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from predictive_maintenance.data_generator import SyntheticDataGenerator

%matplotlib inline
sns.set_style('whitegrid')
```

## 1. Generate Synthetic Data

```python
# Initialize data generator
generator = SyntheticDataGenerator(random_state=42)

# Generate complete dataset
train_df, test_df = generator.generate_complete_dataset(n_units=50)

print(f"Training set: {train_df.shape}")
print(f"Test set: {test_df.shape}")
```

## 2. Explore the Data

```python
# Display first few rows
train_df.head()
```

```python
# Check data types and missing values
print(train_df.info())
print("\nMissing values:")
print(train_df.isnull().sum())
```

## 3. Visualize RUL Distribution

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(train_df['RUL'], bins=50, edgecolor='black')
plt.xlabel('Remaining Useful Life (cycles)')
plt.ylabel('Frequency')
plt.title('RUL Distribution')

plt.subplot(1, 2, 2)
train_df['will_fail'].value_counts().plot(kind='bar')
plt.xlabel('Failure Label')
plt.ylabel('Count')
plt.title('Failure Label Distribution')
plt.xticks([0, 1], ['No Failure', 'Failure'], rotation=0)

plt.tight_layout()
plt.show()
```

## 4. Visualize Sensor Degradation Over Time

```python
# Select a unit to visualize
sample_unit = train_df[train_df['unit_id'] == 1].copy()

fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Plot sensor readings
axes[0, 0].plot(sample_unit['cycle'], sample_unit['sensor_1'], label='Sensor 1')
axes[0, 0].plot(sample_unit['cycle'], sample_unit['sensor_2'], label='Sensor 2')
axes[0, 0].set_xlabel('Cycle')
axes[0, 0].set_ylabel('Sensor Value')
axes[0, 0].set_title('Sensor Readings Over Time')
axes[0, 0].legend()

# Plot operational settings
axes[0, 1].plot(sample_unit['cycle'], sample_unit['temperature'], label='Temperature')
axes[0, 1].set_xlabel('Cycle')
axes[0, 1].set_ylabel('Temperature')
axes[0, 1].set_title('Temperature Over Time')

axes[1, 0].plot(sample_unit['cycle'], sample_unit['vibration'], label='Vibration', color='red')
axes[1, 0].set_xlabel('Cycle')
axes[1, 0].set_ylabel('Vibration')
axes[1, 0].set_title('Vibration Over Time')

axes[1, 1].plot(sample_unit['cycle'], sample_unit['RUL'], label='RUL', color='green')
axes[1, 1].set_xlabel('Cycle')
axes[1, 1].set_ylabel('RUL')
axes[1, 1].set_title('Remaining Useful Life Over Time')

plt.tight_layout()
plt.show()
```

## 5. Correlation Analysis

```python
# Select numeric columns for correlation
numeric_cols = ['sensor_1', 'sensor_2', 'sensor_3', 'temperature', 'pressure', 'vibration', 'RUL']
correlation_matrix = train_df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Feature Correlation Matrix')
plt.show()
```

## 6. Save Datasets

```python
# Save to CSV
train_df.to_csv('../data/train_data.csv', index=False)
test_df.to_csv('../data/test_data.csv', index=False)

print("Datasets saved successfully!")
```
